## Imports

```python
from KeywordDetection import RabinKarpKeywordScanner, summarize_results
from pathlib import Path
import json
```

## Load Keywords

```python
scanner = RabinKarpKeywordScanner(case_sensitive=False)

scanner.add_keywords_from_file("SampleKeywords/keywords_finance.txt")
scanner.add_keywords_from_file("SampleKeywords/keywords_technology.txt")
```

## Verify Keywords Loaded

```python
scanner._pattern_set
```

## Finance Company Example

```python
finance_email_files = list(Path("SampleEmails/Finance").glob("*.txt"))

finance_results = {
    str(path): scanner.scan_file(path)
    for path in finance_email_files
}
```

```python
finance_summary = summarize_results(finance_results)
finance_summary
```

```python
print(json.dumps(finance_summary, indent=2))
```
